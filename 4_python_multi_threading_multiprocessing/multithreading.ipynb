## now with the help of the multithreading we try to reduce the time

```python
import threading
import time

def print_numbers():
    
    for i in range(5):
        time.sleep(1)
        print(i)

def print_letters():
    
    for letter in 'abcdef':
        time.sleep(1)
        print(letter)

## creating 2 threads
thread_1 = threading.Thread(target = print_numbers)
thread_2 = threading.Thread(target = print_letters)


t = time.time()


## start the thread
thread_1.start()
thread_2.start()


## wait for the thread to continue
thread_1.join()
thread_2.join()


time_taken = time.time()-t
print(time_taken)
```
