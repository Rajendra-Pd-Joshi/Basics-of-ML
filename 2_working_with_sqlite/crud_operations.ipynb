```python
import sqlite3
```

```python
## Connect to an sqlite database
connection = sqlite3.connect('example.db')
connection
```

```python
cursor = connection.cursor()
```

```python
## Create a table
cursor.execute('''
Create Table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
     department text
    )
''')

## commit the changes
connection.commit()
```

```python
## Insert the data in sqlite table
cursor.execute('''
insert into employees(name,age,department)
    values('Raju','23','agentic ai')
''')
cursor.execute('''
insert into employees(name,age,department)
    values('dinesh','21','doctor')
''')
cursor.execute('''
insert into employees(name,age,department)
    values('jayanti','20','army')
''')
## commit the changes
connection.commit()
```

```python
## query the data from the table
cursor.execute('''
select * from employees
''')

rows = cursor.fetchall()

## print the queried data

for row in rows:
    print(row)

```

```python
## update the data in the table
cursor.execute('''
update employees
Set age = 34 
where name = "Raju"
''')

connection.commit()
```

```python
cursor.execute('''
select * from employees
''')

rows = cursor.fetchall()

## print the queried data

for row in rows:
    print(row)
```

```python
## delete the data from the table
cursor.execute('''
delete from employees
where name = "jayanti"
''')

connection.commit()
```

```python
cursor.execute('''
select * from employees
''')

rows = cursor.fetchall()

## print the queried data

for row in rows:
    print(row)
```

```python
## close the connection
connection.close()
```
